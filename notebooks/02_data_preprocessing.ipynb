```python
import pandas as pd

df = pd.read_csv("../data/raw/logon.csv")

print(df.shape)
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")

print(df["date"].min())
print(df["date"].max())
```

```python
df = df.dropna(subset=["user", "pc", "date", "activity"])
```

```python
df = df.drop_duplicates()
```

```python
df = df.sort_values(["user", "date"]).reset_index(drop=True)
```

```python
print(df["activity"].value_counts())
```

```python
df["hour"] = df["date"].dt.hour
df["day"] = df["date"].dt.date
df["weekday"] = df["date"].dt.dayofweek
```

```python
df.to_csv("../data/processed/logon_cleaned.csv", index=False)

print("Saved cleaned dataset.")
print(df.shape)
```
